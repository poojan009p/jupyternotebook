# shortcuts practice of jupyter notebook

| Shortcut      | Action                       |
| ------------- | ---------------------------- |
| Shift + Enter | Run cell and move to next    |
| Ctrl + Enter  | Run cell and stay            |
| Alt + Enter   | Run cell and create new cell |
| Esc           | Command mode                 |
| Enter         | Edit mode                    |

## Moving Around Notebook

Move cell up - K

Move cell down - j

Move selected cell up - shift + K

Move selected cell down - shift + j
